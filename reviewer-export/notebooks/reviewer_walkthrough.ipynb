# Reviewer walkthrough

This notebook loads the CSV files emitted by the reviewer-export pipeline (Phase 11) and shows simple counts + previews. It is reviewer-safe:

- it uses only the Python standard library;
- pandas and matplotlib are imported behind guards (the notebook still works without them);
- it never reaches the network;
- it does not require ROS, Gazebo, or Foxglove.

_The platform is **not safety-certified**; this notebook is engineering review material._

CSV files are expected at ``../csv/<table>.csv``.


```python
import csv
import json
from pathlib import Path

CSV_DIR = Path('../csv').resolve()
print('CSV directory:', CSV_DIR)
print('Files:', sorted(p.name for p in CSV_DIR.glob('*.csv')))

try:
    import pandas as _pd  # type: ignore
    _HAS_PANDAS = True
except Exception:
    _HAS_PANDAS = False

def load_table(name):
    path = CSV_DIR / f'{name}.csv'
    if not path.is_file():
        return []
    if _HAS_PANDAS:
        return _pd.read_csv(path)
    with path.open('r', encoding='utf-8') as fh:
        return list(csv.DictReader(fh))

def preview(rows, limit=5):
    if _HAS_PANDAS and hasattr(rows, 'head'):
        return rows.head(limit)
    return rows[:limit]
```

## Programme health

```python
rows = load_table('programme_health')
print('row count:', len(rows) if not _HAS_PANDAS or not hasattr(rows, 'shape') else rows.shape[0])
preview(rows)
```

## How to interpret this export

- Static-only evidence stays static-only across the export.   The replay-quality table records the original ``static_only``   flag verbatim; never read it as live evidence.
- Missing-bag evidence stays missing-bag. The reviewer   cannot assume bag-backed coverage exists.
- ``causality_claimed`` is always ``false`` in subsystem-risk;   the export never claims source-level causation.
- Trend series with ``insufficient_history`` mean the   underlying programme review had fewer than two samples.
